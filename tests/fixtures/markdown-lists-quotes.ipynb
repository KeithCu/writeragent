## Where can I get help?

1. **The cheat sheet** - keep it nearby.
2. **Search for it** - try these:
    * [NumPy documentation](https://numpy.org/doc/stable/index.html) - official docs
    * [Stack Overflow](https://stackoverflow.com/) - questions
    * [ChatGPT](https://chat.openai.com/) - last resort

3. **Ask for help** - after searching.


> **Note:** Important to remember `ndarray` is the core type.
